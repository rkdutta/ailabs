### Decision tree
A Decision Tree is a Flow Chart, and can help you make decisions based on previous experience. Decision Tree gives you different results if you run it enough times, even if you feed it with the same data.

That is because the Decision Tree does not give us a 100% certain answer. It is based on the probability of an outcome, and the answer will vary.


`Important:` To make a decision tree, all data has to be numerical.
We have to convert the non numerical columns from raw data 'Nationality' and 'Go' into numerical values.

```python
import pandas
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt

df = pandas.read_csv("03 - data.csv")

print("RAW DATA")
print(df)

# map nationality to numeric
d = {'UK': 0, 'USA': 1, 'N': 2}
df['Nationality'] = df['Nationality'].map(d)

# map YES/NO to numeric
d = {'YES': 1, 'NO': 0}
df['Go'] = df['Go'].map(d)

# print("\n\nTRANSFORMED DATA")
# print(df)


# Then we have to separate the feature columns from the target column.
# The feature columns are the columns that we try to predict from, and the target column is the column with the values we try to predict.
features = ['Age', 'Experience', 'Rank', 'Nationality']

X = df[features]
y = df['Go']

# init a DecisionTree model
dtree = DecisionTreeClassifier()
dtree = dtree.fit(X, y)

# print decision tree
# tree.plot_tree(dtree, feature_names=features)


# Predict
predictFor = pandas.DataFrame({
    'Age': [40],
    'Experience': [10],
    'Rank': [7],
    'Nationality': [1]
})

print("\n\nShould I go see a show starring a 40 years old American comedian, with 10 years of experience, and a comedy ranking of 7?")
print(dtree.predict(predictFor))
```
