### Canculate mean,median,mode using numpy

````python
# Example 0
import numpy
from scipy import stats

speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

x = numpy.mean(speed)
y = numpy.median(speed)

# The Mode value is the value that appears the most number of times:
# The mode() method returns a ModeResult object that contains the mode number (86), and count (how many times the mode number appeared (3)).
z = stats.mode(speed)

print("mean=",x)
print("median=",y)
print("mode=",z)
````

### Standard Deviation (shortcut σ, pronounce as sigma)
Standard deviation is a number that describes how spread out the values are.

A low standard deviation means that most of the numbers are close to the mean (average) value.

A high standard deviation means that the values are spread out over a wider range.

Example: This time we have registered the speed of 7 cars:

speed = [86,87,88,86,87,85,86]

The standard deviation is: 0.9

Meaning that most of the values are within the range of 0.9 from the mean value, which is 86.4.

````python
# Example 1
import numpy

speed = [86,87,88,86,87,85,86]

print("mean = ",numpy.mean(speed))
print("std. dev = ",numpy.std(speed))
````

````python
# Example 2
import numpy

speed = [32,111,138,28,59,77,97]

print("mean = ",numpy.mean(speed))
print("std. dev = ",numpy.std(speed))
print("Here the std. dev is very high means the values are scattered from the mean line")
````

### Variance (σ2)
Variance is another number that indicates how spread out the values are.

In fact, if you take the square root of the variance, you get the standard deviation!

Or the other way around, if you multiply the standard deviation by itself, you get the variance!

To calculate the variance you have to do as follows:

````python
# Example 3
import numpy

speed = [32,111,138,28,59,77,97]

print("mean = ",numpy.mean(speed))
print("std. dev = ",numpy.std(speed))
print("variance = ",numpy.var(speed))
print("\nImportant: variance = Square root of Std. dev ")
variance = numpy.var(speed)
print("variance = ",variance)
print("sqrt of variance = ",numpy.sqrt(variance),"(this is same as std. dev)")
````

### Normal Data Distribution

An array where the values are concentrated around a given value. 

In probability theory this kind of data distribution is known as the normal data distribution, or the Gaussian data distribution.

A normal distribution graph is also known as the bell curve because of it's characteristic shape of a bell.

````python
import numpy
import matplotlib.pyplot as plt

# Difference between this and revious example is here we have used normal function instead of uniform function
x = numpy.random.normal(5.0, 1.0, 100000) # (mean = 5.0, and standard deviation=1.0, size=100000)

plt.hist(x, 100)
plt.show()
````

### Regression
The term regression is used when you try to find the relationship between variables where the values can take a wide range of values.

In Machine Learning, and in statistical modeling, that relationship is used to predict the outcome of future events.

`Regression` predicts a number from infinitly many possible outputs.

`Linear regression` uses the relationship between the data-points to draw a straight line through all them.

````python
# Data set for linear regression

import matplotlib.pyplot as plt

x = [5,7,8,7,2,17,2,9,4,11,12,9,6] # age of cars
y = [99,86,87,88,111,86,103,87,94,78,77,85,86] # speed of cars

plt.scatter(x, y)
plt.xlabel("age of cars")
plt.ylabel("speed of cars")
plt.show()
````

````python
# Linear Regression

import matplotlib.pyplot as plt
from scipy import stats

# Here, we want to see if y can be predicted from x
x = [5,7,8,7,2,17,2,9,4,11,12,9,6]
y = [99,86,87,88,111,86,103,87,94,78,77,85,86]

# linregress function fits a line
# y = slope * x + intercept
# It calculates the best-fit line using least squares.
slope, intercept, r, p, std_err = stats.linregress(x, y)

#| Variable    | Meaning                                                                  
#| ----------- | --------------------------------------------------------------------------------------
#| `slope`     | How much `y` changes for a unit change in `x` (rise/run)                 
#| `intercept` | Value of `y` when `x = 0` (where line crosses y-axis)                    
#| `r`         | Correlation coefficient (strength of linear relationship, -1 to +1. 0=no relationship
#| `p`         | p-value → probability that slope is zero                                 
#| `std_err`   | Standard error of the estimated slope (uncertainty)                      


# model is function of x i.e. f(x).
# Create a function that uses the slope and intercept values to return a new value. This new value represents where on the y-axis the corresponding x value will be placed.
def f(x):
  return slope * x + intercept

# Run each value of the x array through the function. This will result in a new array with new values for the y-axis.
mymodel = list(map(f, x))

plt.scatter(x, y,label="Data points")
plt.plot(x, mymodel,color='red',label="Best fit line")

# Predicted speed for a 10 years old car
x = 10
speed = f(x)
plt.scatter(x, speed, color='green', s=100, label=f"Predicted speed of 10 yrs: {speed:.2f}")


plt.xlabel("age of cars")
plt.ylabel("speed of cars")
plt.legend()
plt.show()
print("r=",r, "# shows the relationship strength")
````

````python
# Bad fit

import matplotlib.pyplot as plt
from scipy import stats

x = [89,43,36,36,95,10,66,34,38,20,26,29,48,64,6,5,36,66,72,40]
y = [21,46,3,35,67,95,53,72,58,10,26,34,90,33,38,20,56,2,47,15]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
  return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.show()

print("r=",r," # value is close to zero means almost no relationship to the datapoints")
print("Linear regression is not a good strategy for this data points. If your data points clearly will not fit a linear regression (a straight line through all data points), it might be ideal for polynomial regression.")
````

### Polynomial Regression

Polynomial regression, like linear regression, uses the relationship between the variables x and y to find the best way to draw a line through the data points.

In the example below, we have registered 18 cars as they were passing a certain tollbooth.
We have registered the car's speed, and the time of day (hour) the passing occurred. The x-axis represents the hours of the day and the y-axis represents the speed.

It is important to know how well the relationship between the values of the x- and y-axis is, if there are no relationship the polynomial regression can not be used to predict anything. The relationship is measured with a value called the r-squared.

````python
import numpy
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

x = [1,2,3,5,6,7,8,9,10,12,13,14,15,16,18,19,21,22]
y = [100,90,80,60,60,55,60,65,70,70,75,76,78,79,90,99,99,100]

# This line creates a polynomial regression model of degree 3 (a cubic curve)
# y=ax3+bx2+cx+d
mymodel = numpy.poly1d(numpy.polyfit(x, y, 3))
# mymodel becomes a mathematical function representing the cubic regression curve.

# This creates 100 evenly spaced numbers between 1 and 22.
myline = numpy.linspace(1, 22, 100)

plt.scatter(x, y)
plt.plot(myline, mymodel(myline),label="Best fit line")

# Predicted speed at 17:00 hrs
x1 = 17
speed = mymodel(17)
plt.scatter(x1, speed, color='green', s=100, label=f"Predicted speed at 5 PM: {speed:.2f}")


plt.xlabel("hour of day")
plt.ylabel("speed of cars")
plt.legend()
plt.show()
print("r2_score=",r2_score(y, mymodel(x)))
print("The result 0.94 shows that there is a very good relationship, and we can use polynomial regression in future predictions.")
````

````python
# Bad fit for polynomial regression
# # same example different values

import numpy
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score

x = [89,43,36,36,95,10,66,34,38,20,26,29,48,64,6,5,36,66,72,40]
y = [21,46,3,35,67,95,53,72,58,10,26,34,90,33,38,20,56,2,47,15]

# This line creates a polynomial regression model of degree 3 (a cubic curve)
# y=ax3+bx2+cx+d
mymodel = numpy.poly1d(numpy.polyfit(x, y, 3))
# mymodel becomes a mathematical function representing the cubic regression curve.

# This creates 100 evenly spaced numbers between 1 and 22.
myline = numpy.linspace(1, 22, 100)

plt.scatter(x, y)
plt.plot(myline, mymodel(myline),label="Best fit line")

# Predicted speed at 17:00 hrs
x1 = 17
speed = mymodel(17)
plt.scatter(x1, speed, color='green', s=100, label=f"Predicted speed at 5 PM: {speed:.2f}")


plt.xlabel("hour of day")
plt.ylabel("speed of cars")
plt.legend()
plt.show()
r2_score = r2_score(y, mymodel(x))
print("r2_score=",r2_score)
print(f"The result {r2_score} shows very bad relationship.")
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from mpl_toolkits.mplot3d import Axes3D

# Load dataset
df = pd.read_csv("01 - data.csv")

# print(df)

# Features and target
X = df[['Weight', 'Volume']]
y = df['CO2']


# Train model
regr = LinearRegression()
regr.fit(X, y)


# The coefficient is a factor that describes the relationship with an unknown variable.
# Example: if x is a variable, then 2x is x two times. x is the unknown variable, and the number 2 is the coefficient.
# In this case, we can ask for the coefficient value of weight against CO2, and for volume against CO2. The answer(s) we get tells us what would happen if we increase, or decrease, one of the independent values.
print("\n\nCoefficient of the trained model = ", regr.coef_)

print(f"These values tell us that if the weight increase by 1kg, the CO2 emission increases by {regr.coef_[0]}g.")

print(f"And if the engine size (Volume) increases by 1cm3, the CO2 emission increases by {regr.coef_[1]}g.")

# -----------------------------
# Prediction for new car
# -----------------------------
new_car = pd.DataFrame({
    'Weight': [2300],
    'Volume': [1300]
})

print("\n\npredict CO2 emmission for this new car:\n",new_car)
predicted_co2 = regr.predict(new_car)[0]   # <-- defined here
print(f"Predicted CO2 emission: {predicted_co2:.2f}")

# -----------------------------
# 3D Plot
# -----------------------------
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Scatter actual data
ax.scatter(
    df['Weight'],
    df['Volume'],
    df['CO2'],
    color='blue',
    label="Actual Data"
)

# Create grid for regression plane
weight_range = np.linspace(df['Weight'].min(), df['Weight'].max(), 30)
volume_range = np.linspace(df['Volume'].min(), df['Volume'].max(), 30)
weight_grid, volume_grid = np.meshgrid(weight_range, volume_range)

grid_df = pd.DataFrame({
    'Weight': weight_grid.ravel(),
    'Volume': volume_grid.ravel()
})

co2_pred = regr.predict(grid_df)
co2_pred = co2_pred.reshape(weight_grid.shape)

# Plot plane
ax.plot_surface(weight_grid, volume_grid, co2_pred, alpha=0.3, color='red')

# Plot predicted point
ax.scatter(
    2300,
    1300,
    predicted_co2,
    color='green',
    s=100,
    label=f"Prediction\nCO2={predicted_co2:.2f}"
)

ax.set_xlabel("Weight")
ax.set_ylabel("Volume")
ax.set_zlabel("CO2")
plt.legend()
plt.show()
````

### Scaling

If a dataset contains data from various ranges, then it is first of difficult to plot for normal scaling reasons, secondly it can be very resource intensive to train and predict.

When your data has different values, and even different measurement units, it can be difficult to compare them. What is kilograms compared to meters? Or altitude compared to time?

The answer to this problem is scaling. We can scale data into new values that are easier to compare.

It can be difficult to compare the volume 1.0 with the weight 790, but if we scale them both into comparable values, we can easily see how much one value is compared to the other.

The standardization method uses this formula:

z = (x - u) / s

Where z is the new value, x is the original value, u is the mean and s is the standard deviation.
Python sklearn module has a method called `StandardScaler()` which returns a Scaler object with methods for transforming data sets.

````python
import pandas
from sklearn import linear_model
from sklearn.preprocessing import StandardScaler
scale = StandardScaler()

df = pandas.read_csv("01 - data.csv")

X = df[['Weight', 'Volume']]
y = df['CO2']

scaledX = scale.fit_transform(X)

regr = linear_model.LinearRegression()
regr.fit(scaledX, y)

new_car = pandas.DataFrame({
    'Weight': [2300],
    'Volume': [1.3]
})

scaled = scale.transform(new_car)
predictedCO2 = regr.predict(scaled)

print(predictedCO2)
````
