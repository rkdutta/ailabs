### Canculate mean,median,mode using numpy

```python
# Example 0
import numpy
from scipy import stats

speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

x = numpy.mean(speed)
y = numpy.median(speed)

# The Mode value is the value that appears the most number of times:
# The mode() method returns a ModeResult object that contains the mode number (86), and count (how many times the mode number appeared (3)).
z = stats.mode(speed)

print("mean=",x)
print("median=",y)
print("mode=",z)
```

### Standard Deviation (shortcut σ, pronounce as sigma)
Standard deviation is a number that describes how spread out the values are.

A low standard deviation means that most of the numbers are close to the mean (average) value.

A high standard deviation means that the values are spread out over a wider range.

Example: This time we have registered the speed of 7 cars:

speed = [86,87,88,86,87,85,86]

The standard deviation is: 0.9

Meaning that most of the values are within the range of 0.9 from the mean value, which is 86.4.

```python
# Example 1
import numpy

speed = [86,87,88,86,87,85,86]

print("mean = ",numpy.mean(speed))
print("std. dev = ",numpy.std(speed))
```

```python
# Example 2
import numpy

speed = [32,111,138,28,59,77,97]

print("mean = ",numpy.mean(speed))
print("std. dev = ",numpy.std(speed))
print("Here the std. dev is very high means the values are scattered from the mean line")
```

### Variance (σ2)
Variance is another number that indicates how spread out the values are.

In fact, if you take the square root of the variance, you get the standard deviation!

Or the other way around, if you multiply the standard deviation by itself, you get the variance!

To calculate the variance you have to do as follows:

```python
# Example 3
import numpy

speed = [32,111,138,28,59,77,97]

print("mean = ",numpy.mean(speed))
print("std. dev = ",numpy.std(speed))
print("variance = ",numpy.var(speed))
print("\nImportant: variance = Square root of Std. dev ")
variance = numpy.var(speed)
print("variance = ",variance)
print("sqrt of variance = ",numpy.sqrt(variance),"(this is same as std. dev)")
```
