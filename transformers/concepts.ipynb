# Transformers

Transformers are a neural network architecture that handles sequential data as input and outputs sequential data. They convert one sequence to another — e.g. Q&A systems, text summarization, and translation.

## Architecture

The Transformer architecture, introduced in ["Attention Is All You Need"](https://arxiv.org/abs/1706.03762) (Vaswani et al., 2017):

![Transformer Architecture](https://upload.wikimedia.org/wikipedia/commons/8/8f/The-Transformer-model-architecture.png)

The model consists of an **encoder** (left) and **decoder** (right), both built from stacked self-attention and feed-forward layers.

Word Embeddings -- Embeddings are learned from a vocabulary of words. They are used to represent words as vectors in a high-dimensional space.

For example, the word "the" might be represented as a vector of 1000 dimensions, with the first 999 dimensions being 0 and the last dimension being 1. This is a simple example, but it illustrates the idea of word embeddings.


For example, Googles word2vec model is represented by a 300-dimensional vector.

Following is an simple example or generating embeddings. There are several different techniques.

```python
from sklearn.feature_extraction.text import CountVectorizer
documents = ["This is the first document.", "This document is the second document.",
              "And this is the third one.", "cat dog", "Is this the first document?"]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)
feature_names = vectorizer.get_feature_names_out()

print("Bag-of-Words Matrix:")
print(X.toarray())
print('Dimension =',X.toarray()[0].size)
print("Vocabulary (Feature Names):", feature_names)
```
