## workflow vs agents

In a `workflow`, a developer writes the code that decides the order of operations — the LLM is just one step in a predetermined sequence. It's predictable and fast, but rigid.

In an `agent`, the LLM itself decides what tools to call, in what order, and whether to keep going or stop. It can adapt mid-task, loop back, or take paths the developer never explicitly programmed. That's the power — and also the unpredictability — of agents.

![alt text](image.png)

https://www.anthropic.com/engineering/building-effective-agents

> **Agents run tools in a loop to achieve a goal.**
Applications that integrate LLMs but don’t use them to control workflow execution—think simple
chatbots, single-turn LLMs, or sentiment classifiers—`are not agents`.

## when should you consider building an agent?

1. **Complex decision-making**: Workflows involving nuanced judgment, exceptions, or 
context-sensitive decisions, for example refund approval in customer service workflows.

2. **Difficult-to-maintain rules**: Systems that have become unwieldy due to extensive and
intricate rulesets, making updates costly or error-prone,for example performing vendor security reviews. 

3. **Heavy reliance on unstructured data**: Scenarios that involve interpreting natural language, extracting meaning from documents, or interacting with users conversationally, for example processing a home insurance claim. 

Components:

1. LLM (Brain) - Reasoning, planning, decision making
2. Tools - hands (web search, read files, write code, access dbs)
3. Instructions - Explicit guidelines and guardrails defining how the agent behaves

## Claude Code vs Agent SDK — Side by Side

There are two ways how you can define an agent. The decision depends on the question "Who is the orchestrator?"

### Core Difference

**Claude Code** — you are the orchestrator. You type prompts in a terminal, 
Claude responds, asks for your approval before editing files, and you guide 
the session interactively. It's a conversation with an agent.

**Agent SDK** — your Python script is the orchestrator. You write code that 
programmatically launches the agent, defines its tools and permissions, and 
processes its output. The agent runs autonomously without a human in the loop.

> You will need an API key to use Agent SDK.

---

### How the flow looks

**Claude Code**
```
You → "Review utils.py and fix bugs"
Claude → reads file... proposes fixes... asks permission... edits
You → approve/reject each step
```

**Agent SDK**
```
Your script → launches agent → agent runs to completion → returns results
              (no human involved mid-run)
```

---

### Agent SDK Code Example
```python
async for message in query(
    prompt="Review utils.py for bugs and fix them",
    options=ClaudeAgentOptions(
        allowed_tools=["Read", "Edit", "Glob"],
        permission_mode="acceptEdits",  # no human approval needed
    ),
):
    print(message)
```

### Comparison Table

| | Claude Code | Agent SDK |
|---|---|---|
| Who drives it | You, interactively | Your Python script |
| Needs API key | No (Pro login) | Yes |
| Human approval | Yes, at each step | Optional / configurable |
| Can embed in apps | No | Yes |
| Good for learning | ✅ Best starting point | After you understand agents |
| Good for automation | Not really | ✅ CI/CD, scheduled jobs, products |
| Where agent "lives" | Your terminal session | Inside your codebase |

---

### Key Insight

Both use the **exact same agent loop** under the hood — Claude reading files, 
deciding what to do, calling tools, looping until done. The difference is 
purely about who or what is sitting in the driver's seat.

Claude Code is the SDK with a human in the loop. The Agent SDK is Claude Code 
with a Python script in the loop instead.
